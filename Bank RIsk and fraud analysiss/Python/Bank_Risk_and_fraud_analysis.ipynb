Q6 - Customer spending analysis
* Top 10 customer by total transaction
* Bottom 10 customer by total transaction
* Average spending per customer

```python
# Top 10 customer by total transaction

top_10_customer = df.groupby("AccountID")["TransactionAmount"].sum().sort_values(ascending=False).head(10)
print(top_10_customer)
```

```python
# Bottom 10 customer by total transaction

bottom10customer = df.groupby("AccountID")["TransactionAmount"].sum().sort_values(ascending=False).tail(10)
print(bottom10customer)
```

```python
# Average spending per customer

averagespending = (
    df.groupby('AccountID',as_index=False)['TransactionAmount'].mean()
)
print(averagespending)
```

Q7 - Merchant Performance analysis

* Top Merchant by Transaction value
* Top Merchant by Transaction count
* Average Transaction per merchant

```python
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
# Top Merchant by Transaction value.
top_merchant_value = df.groupby("MerchantID")["TransactionAmount"].sum().sort_values(ascending=False).head(5)
print(top_merchant_value)
```

```python
top_merchant_value.plot(kind="bar")
plt.title("Top Merchant by transaction value")
plt.xlabel("MerchantID")
plt.ylabel("Transaction amount")
plt.show()
```

```python
# Top 5 Merchant by Transaction count
top_merchant_count = df.groupby("MerchantID")["TransactionID"].count().sort_values(ascending=False).head(5)
print(top_merchant_count)
```

```python
top_merchant_count.plot(kind="bar")
plt.title("Top 5 merchant count")
plt.xlabel("MerchantID")
plt.ylabel("TransactionID")
plt.show()
```

```python
# Average Transaction per merchant

average_transaction_per_merchant = df.groupby("MerchantID")["TransactionAmount"].mean()
print(average_transaction_per_merchant)
```

```python
average_transaction_per_merchant.plot(kind="line")
plt.title("Average transaction per merchant")
plt.xlabel("MerchantID")
plt.ylabel("TransactionAmount")
plt.show()
```

Q10 - Time Based Analysis
* Transaction per day
* Transaction per month
* Monthly revenue
* Daily average transaction amount

```python
# Transaction per day

df["TransactionDate"] = pd.to_datetime(df["TransactionDate"])

transactionperday = (
    df.groupby(df["TransactionDate"].dt.date)["TransactionID"].count().reset_index(name="TransactionCount")
)

print(transactionperday)
```

```python
transactionperday.plot(kind="line")
plt.title("Transaction per day")
plt.xlabel("TransactionDate")
plt.ylabel("TransactionID")
plt.show()
```

```python
# Transaction per month

df["TransactionDate"] = pd.to_datetime(df["TransactionDate"])

transactionpermonth = (
    df.groupby(df["TransactionDate"].dt.month)["TransactionID"].count().reset_index(name="TransactionCount")
)

print(transactionpermonth)
```

```python
transactionpermonth.plot(kind="bar")
plt.title("Transacton per month")
plt.show()
```
